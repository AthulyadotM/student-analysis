```python
import pandas as pd
```

```python
df=pd.read_csv("student_performance_500.csv")
df
```

```python
df.isnull().values.any()
```

```python
df=df.drop_duplicates()
```

```python
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
df["Final_Mark"] = df["Final_Mark"].fillna(df["Final_Mark"].mean())
```

```python
df
```

```python
df.isnull().sum()
```

```python
low_attendance = df[df["Attendance"] < 75]
low_attendance
```

```python
high_scorers = df[df["Final_Mark"] > 80]
high_scorers
```

```python
passed = df[df["Result"] == "Pass"]
passed
```

```python
df["Total_Score"] = (
    df["Assignment_Score"] +
    df["Internal_Mark"] +
    df["Final_Mark"]
)
df
```

```python
dept_average = df.groupby("Department")["Final_Mark"].mean()
dept_average
```

```python
gender_average = df.groupby("Gender")["Final_Mark"].mean()
gender_average
```

```python
df_sorted = df.sort_values("Final_Mark", ascending=False)
df_sorted
```

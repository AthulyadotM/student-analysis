```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Histogram/KDE

Box plot

Count plot

Scatter plot

Heatmap showing correlations

```python
df=pd.read_csv("student_performance_500.csv")
```

```python
sns.histplot(data=df, x="Final_Mark", bins=10,kde=True)
plt.title("Department-wise Average Mark")
plt.show()
```

```python
sns.boxplot(data=df, x="Department", y="Final_Mark")
plt.title("Final Marks by Department")
plt.show()
```

```python
sns.countplot(data=df, x="Department")
plt.title("Number of Students by Department")
```

```python
sns.scatterplot(data=df, x="Study_Hours", y="Final_Mark")
plt.title("Study_Hours vs Final_Mark")
```

```python
corr = df.select_dtypes("number").corr()
sns.heatmap(corr, annot=True)
plt.show()
```

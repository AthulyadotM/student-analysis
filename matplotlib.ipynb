```python
import pandas as pd
df=pd.read_csv("student_performance_500.csv")
df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["Final_Mark"], bins=10)
plt.xlabel("Final Mark")
plt.ylabel("Number of Students")
plt.title("Student Mark Distribution")
plt.show()
```

```python
dept_avg = df.groupby("Department")["Final_Mark"].mean()
plt.figure(figsize=(8, 5))
plt.bar(dept_avg.index, dept_avg.values)
plt.xlabel("Department")
plt.ylabel("Average Final Mark")
plt.title("Department-wise Average Final Marks")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Attendance"], bins=10)
plt.xlabel("Attendance")
plt.ylabel("Number of Students")
plt.title("Student Attendance Distribution")
plt.show()
```

```python
result_count = df["Result"].value_counts()
plt.figure(figsize=(6, 5))
plt.bar(result_count.index, result_count.values)
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.title("Pass vs Fail")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Study_Hours"], df["Final_Mark"])
plt.xlabel("Study Hours")
plt.ylabel("Final Mark")
plt.title("Study Hours vs Final Marks")
plt.show()
```

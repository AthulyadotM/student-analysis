```python
import numpy as np
import pandas as pd
```

Mean

Median

Standard deviation

Minimum/maximum

Percentiles

Correlation between selected numerical variables

```python
df=pd.read_csv("student_performance_500.csv")
```

```python
mean_mark = np.mean(df["Final_Mark"])
print("Mean:", mean_mark)
```

```python
median_mark = np.median(df["Final_Mark"])
print("Median:", median_mark)
```

```python
std_mark = np.std(df["Final_Mark"])
print("Standard Deviation:", std_mark)
```

```python
print("Minimum:", np.min(df["Final_Mark"]))
print("Maximum:", np.max(df["Final_Mark"]))
```

```python
print("25th Percentile:", np.percentile(df["Final_Mark"], 25))
print("50th Percentile:", np.percentile(df["Final_Mark"], 50))
print("75th Percentile:", np.percentile(df["Final_Mark"], 75))
```

```python
correlation = np.corrcoef(
    df["Study_Hours"],
    df["Final_Mark"]
)

print(correlation)
```
